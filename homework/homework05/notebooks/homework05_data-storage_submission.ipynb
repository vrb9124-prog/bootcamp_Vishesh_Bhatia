```python
import sys
sys.path.append('../src')
from storage import write_df, read_df
from dotenv import load_dotenv
import os
import pandas as pd
from datetime import datetime

load_dotenv()
raw_dir = os.getenv('DATA_DIR_RAW')
processed_dir = os.getenv('DATA_DIR_PROCESSED')
print("Raw dir:", raw_dir)
print("Processed dir:", processed_dir)
```

```python
df = pd.DataFrame({
    'category': ['A', 'B', 'A', 'B', 'A'],
    'value': [10, 15, 12, 18, 9],
    'date': pd.date_range('2025-08-01', periods=5)
})
df
```

```python
timestamp = datetime.now().strftime('%Y%m%d-%H%M')
csv_path = f"{raw_dir}/sample_{timestamp}.csv"
parquet_path = f"{processed_dir}/sample_{timestamp}.parquet"

write_df(df, csv_path)
write_df(df, parquet_path)
```

```python
df_csv = read_df(csv_path)
df_parquet = read_df(parquet_path)

def validate_roundtrip(original, reloaded, name):
    shape_match = original.shape == reloaded.shape
    print(f"[{name}] Shape match: {shape_match} (original {original.shape}, reloaded {reloaded.shape})")
    print(f"[{name}] Dtypes:\n{reloaded.dtypes}\n")

validate_roundtrip(df, df_csv, "CSV")
validate_roundtrip(df, df_parquet, "Parquet")
```
