## 1. NumPy Operations

```python
# TODO: Create a NumPy array and perform vectorized operations

arr_0 = np.array([i for i in range(100)])
arr_1 = np.array([i/10 for i in range(100)])

#Vectorized Operations:

start_vec_addn=time.perf_counter()
sum_arr = arr_0 + arr_1
end_vec_addn=time.perf_counter()

start_vec_sub=time.perf_counter()
sub_arr = arr_0 - arr_1
end_vec_sub=time.perf_counter()

start_vec_prod=time.perf_counter()
prod_arr = arr_0 * arr_1
end_vec_prod=time.perf_counter()

start_vec_div=time.perf_counter()
div_arr = arr_0[1:]/arr_1[1:]
end_vec_div=time.perf_counter()


#Loops:
start_loop_addn = time.perf_counter()
sum_arr_loop = [arr_0[i] + arr_1[i] for i in range(10)]
end_loop_addn=time.perf_counter()

start_loop_sub = time.perf_counter()
sub_arr_loop = [arr_0[i] - arr_1[i] for i in range(10)]
end_loop_sub=time.perf_counter()

start_loop_prod = time.perf_counter()
prod_arr_loop = [arr_0[i] * arr_1[i] for i in range(10)]
end_loop_prod=time.perf_counter()

start_loop_div = time.perf_counter()
div_arr_loop = [arr_0[i]/arr_1[i] for i in range(1,10)]
end_loop_div=time.perf_counter()

```

## 2. Compare Loop vs Vectorization

```python
print(f"Vectorized Addition Time: {end_vec_addn-start_vec_addn}")
print(f"Vectorized Subtraction Time: {end_vec_sub-start_vec_sub}")
print(f"Vectorized Multiplication Time: {end_vec_prod-start_vec_prod}")
print(f"Vectorized Division Time: {end_vec_div-start_vec_div}")

print(f"Loop Addition Time: {end_loop_addn-start_loop_addn}")
print(f"Loop Subtraction Time: {end_loop_sub-start_loop_sub}")
print(f"Loop Multiplication Time: {end_loop_prod-start_loop_prod}")
print(f"Loop Division Time: {end_loop_div-start_loop_div}")
```
