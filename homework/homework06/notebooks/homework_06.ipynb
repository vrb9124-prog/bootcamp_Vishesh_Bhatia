```python
import sys
sys.path.append('../src')
from cleaning import fill_missing_median, drop_missing, normalize_data
import pandas as pd
import glob
import os


raw_file = glob.glob('../data/raw/*.csv')[0]
print("Using:", raw_file)
df = pd.read_csv(raw_file)
df.head()
```

```python
print("Shape:", df.shape)
print("\nMissing values per column:\n", df.isna().sum())
df.describe()
```

```python
# Step A: fill missing numeric values with the median
df_filled = fill_missing_median(df)
print("Missing after fill:\n", df_filled.isna().sum())
```

```python
# Step B: drop any remaining rows with missing non-numeric data
df_dropped = drop_missing(df_filled)
print("Shape after drop:", df_dropped.shape, "(was", df_filled.shape, ")")
```

```python
# Step C: normalize the numeric columns
df_cleaned = normalize_data(df_dropped)
df_cleaned.describe()
```

```python
comparison = pd.DataFrame({
    'original_shape': [df.shape],
    'cleaned_shape': [df_cleaned.shape],
    'original_missing': [df.isna().sum().sum()],
    'cleaned_missing': [df_cleaned.isna().sum().sum()],
})
comparison
```

## Assumptions
- Missing numeric values are filled with the column median rather than
  dropped, to preserve row count — assumes missingness is not meaningfully
  correlated with the value itself.
- Rows still missing non-numeric fields after that are dropped, since
  there's no sensible way to impute a category.
- Normalization uses min-max scaling — assumes no extreme outliers that
  would distort the 0-1 range (see Stage 07 for outlier handling).

```python
os.makedirs('data/processed', exist_ok=True)
df_cleaned.to_csv('data/processed/cleaned_data.csv', index=False)
print("Saved to data/processed/cleaned_data.csv")
```
